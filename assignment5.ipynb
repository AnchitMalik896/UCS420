```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
max_count = counts.max()

print(values[counts == max_count])

for value in values[counts == max_count]:
    print(np.where(x == value)[0])
```

```python
import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
max_count = counts.max()

print(values[counts == max_count])

for value in values[counts == max_count]:
    print(np.where(y == value)[0])
```

```python
import numpy as np

manya = np.linspace(10, 100, 25)

print(manya)
print(manya.ndim)
print(manya.shape)
print(manya.size)
print(manya.dtype)
print(manya.nbytes)

transpose = manya.reshape(25, 1)
print(transpose)

print(manya.T)
```

```python
import numpy as np

ucs420_manya = np.array([[10, 20, 30, 40],
                         [50, 60, 70, 80],
                         [90, 15, 20, 35]])

print("Mean:", np.mean(ucs420_manya))
print("Median:", np.median(ucs420_manya))
print("Max:", np.max(ucs420_manya))
print("Min:", np.min(ucs420_manya))
print("Unique:", np.unique(ucs420_manya))

reshaped_ucs420_manya = ucs420_manya.reshape(4, 3)
print(reshaped_ucs420_manya)

resized_ucs420_manya = np.resize(ucs420_manya, (2, 3))
print(resized_ucs420_manya)
```
