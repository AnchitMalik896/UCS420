```python
employee_records = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
})

employee_records.to_csv("employees.csv", index=False)

employee_df = pd.read_csv("employees.csv")
print(employee_df)
```

```python
print("Shape :", employee_df.shape)
print("Rows  :", employee_df.shape[0])
print("Columns:", employee_df.shape[1])
```

```python
print("Average salary        :", employee_df["Salary"].mean())
print("Total bonus paid      :", employee_df["Bonus"].sum())
print("Youngest employee age :", employee_df["Age"].min())
print("Highest rating        :", employee_df["Rating"].max())
```

```python
salary_sorted_df = employee_df.sort_values(by="Salary", ascending=False)
print(salary_sorted_df)
```

```python
def categorize(r):
    if r >= 4.5:
        return "Excellent"
    elif r >= 4.0:
        return "Good"
    else:
        return "Average"

employee_df["Performance_Category"] = employee_df["Rating"].apply(categorize)
print(employee_df[["Name", "Rating", "Performance_Category"]])
```

```python
print(employee_df.isnull().sum())
print()
print("Total missing values :", employee_df.isnull().sum().sum())
```

```python
employee_df = employee_df.rename(columns={"Employee_ID": "ID"})
print(employee_df.columns.tolist())
```

```python
print(employee_df[employee_df["Years_of_Experience"] > 5])
print()
print(employee_df[employee_df["Department"] == "IT"])
```

```python
employee_df["Tax"] = employee_df["Salary"] * 0.10
print(employee_df[["Name", "Salary", "Tax"]])
```

```python
employee_df.to_csv("employees_modified.csv", index=False)

verification_df = pd.read_csv("employees_modified.csv")
print(verification_df)
```
