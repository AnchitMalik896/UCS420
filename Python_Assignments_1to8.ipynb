```python
n = int(input("Enter a value for n: "))

running_total = 0
for i in range(1, n + 1):
    running_total = running_total + i

print("Sum from 1 to", n, " --> ", running_total)

print("Sum from 1 to", n, " --> ", sum(range(1, n + 1)))
```

```python
n = int(input("Enter a value for n: "))

divisor_a = 4
divisor_b = 6
total = 0

for i in range(1, n + 1):
    if i % divisor_a == 0 and i % divisor_b == 0:
        total = total + i

print("Sum of numbers divisible by", divisor_a, "and", divisor_b, "--> ", total)
```
