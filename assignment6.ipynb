```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temperature = temperature + 2
print(temperature)

fahrenheit = (9/5) * temperature + 32
print(fahrenheit)

result = temperature[temperature > 32]
print(result)

print(np.sum(temperature > 32))
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print(np.sum(steps))
print(np.mean(steps))
print(np.max(steps))
print(np.min(steps))

print(np.sum(steps, axis=0))
print(np.sum(steps, axis=1))

print(np.argmax(steps))
print(np.unravel_index(np.argmax(steps), steps.shape))
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print(X.shape)
print(X.ndim)

print(X.T)

print(X.T @ X)
```

```python
inverse = np.linalg.inv(X.T @ X)
print(inverse)

beta = inverse @ X.T @ y
print(beta)
```

```python
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

new_user = np.array([5, 40, 7])

prediction = new_user @ beta

print(prediction)
```
